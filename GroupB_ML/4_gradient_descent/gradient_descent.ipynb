# Assignment 4: Gradient Descent Algorithm
Find the local minima of $y = (x + 3)^2$ starting from $x = 2$.

```python
import matplotlib.pyplot as plt
import numpy as np
```

## Function and its derivative

```python
def f(x):
    return (x + 3) ** 2


def df(x):
    return 2 * (x + 3)
```

## Gradient descent implementation

```python
def gradient_descent(start_x, learning_rate=0.1, iterations=100, tolerance=1e-6):
    x = start_x
    history = [x]
    for _ in range(iterations):
        grad = df(x)
        new_x = x - learning_rate * grad
        history.append(new_x)
        if abs(new_x - x) < tolerance:
            x = new_x
            break
        x = new_x
    return x, history
```

## Run from x = 2

```python
start_x = 2
learning_rate = 0.1
minima_x, history = gradient_descent(start_x, learning_rate)

print(f"Converged after {len(history) - 1} iterations")
print(f"Local minima at x = {minima_x:.6f}, y = {f(minima_x):.6f}")
print("Expected minima: x = -3, y = 0")
```

```python
import pandas as pd
pd.DataFrame({"iteration": range(len(history)), "x": history, "y": [f(x) for x in history]}).head(15)
```

## Visualize convergence

```python
xs = np.linspace(-8, 4, 200)
plt.plot(xs, f(xs), label="y = (x+3)^2")
plt.plot(history, [f(x) for x in history], "o-", color="red", label="gradient descent path")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.title("Gradient Descent Convergence")
plt.show()
```
